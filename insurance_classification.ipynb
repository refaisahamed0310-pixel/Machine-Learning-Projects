```python
import pandas as pd
data=pd.read_csv("insurance.csv")
print(data)
```

```python
data.isnull().sum()
```

```python
data.describe()
```

```python
data.info()
```

```python
from sklearn.preprocessing import LabelEncoder
a=LabelEncoder()
data["sex"]=a.fit_transform(data["sex"])
data["smoker"]=a.fit_transform(data["smoker"])
data["region"]=a.fit_transform(data["region"])
```

```python
import seaborn as s
s.barplot(x="sex",y="charges",data=data)
```

```python
import seaborn as s
s.boxplot(x="sex",y="charges",data=data)
```

```python
import seaborn as s
s.lmplot(x="sex",y="charges",data=data)
```

```python
import seaborn as s
s.violinplot(x="sex",y="charges",data=data)
```

```python
import seaborn as s
s.displot(data["charges"])
```

```python
import seaborn as s
s.kdeplot(data["charges"])
```

```python

s.countplot(x=data["charges"])
```

```python
import seaborn as s
s.pairplot(data)
```

```python
b=data.corr(numeric_only=True)
s.heatmap(b,annot=True)
```

```python
x=data.drop("smoker",axis=1)
y=data["smoker"]
print(x)
print(y)
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x, y, test_size=0.20, random_state=42)
print(xtrain)
print(xtest)
print(ytrain)
print(ytest)
```

```python
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
model.fit(xtrain,ytrain)
```

```python
ypred=model.predict(xtest)
print(ypred)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(ytest,ypred))
```

```python
data=model.predict([[    19  ,  0  ,27.900    ,     0     ,  3 , 16884.92400]])
print( data)
```
