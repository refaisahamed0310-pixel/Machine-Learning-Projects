```python
x=data.drop("Legendary",axis=1)
y=data['Legendary']
print(x)
print(y)
```

```python
print(x.columns.size)
print(x)
```

```python
from sklearn.decomposition  import PCA
#PA(principle component analysis)
b=PCA(n_components=8)
# Note: PCA is currently fitted before the train/test split in this original workflow.
# For a leakage-free evaluation, fit PCA on training data only.
x=b.fit_transform(x)


```

```python
print(x)
```

```python
x=pd.DataFrame(data=x,columns=["column1","column2","column3","column4","column5","column6","column7","column8"])
print(x)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x, y, test_size=0.2, random_state=42)
print(x_train)
print(x_test)
print(y_train)
print(y_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
m=RandomForestClassifier()
m.fit(x_train,y_train)
```

```python
ypred=m.predict(x_test)
print(ypred)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,ypred))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,ypred))
```

```python
from sklearn.metrics import confusion_matrix
print(confusion_matrix(y_test,ypred))
```

```python
from sklearn.tree import DecisionTreeClassifier
m=DecisionTreeClassifier()
m.fit(x_train,y_train)
```

```python
ypred=m.predict(x_test)
print(ypred)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,ypred))
```

```python
from sklearn.metrics import confusion_matrix
print(confusion_matrix(y_test,ypred))
```
